# 05 Export Figures

発表用の図を `reports/figures/` に書き出す。

- Ji のヒストグラム
- 寄与3項の平均構成


```python
from __future__ import annotations
from pathlib import Path
import sys

import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import pandas as pd

ROOT = Path.cwd().parents[0] if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))
from heat_town.model import normalize_weights  # noqa: E402

FIG = ROOT / 'reports' / 'figures'
FIG.mkdir(parents=True, exist_ok=True)
```

```python
df = pd.read_parquet(ROOT / 'data' / 'processed' / 'features.parquet')
n1, n2, n3 = normalize_weights(0.3, 0.4, 0.3)
df['ji'] = n1 * df['d'] + n2 * (100 - df['C']) + n3 * df['WBGT']
```

## 図1: Ji の分布

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.hist(df['ji'], bins=30, color='#f4813f', edgecolor='white')
ax.set_xlabel('Ji (低いほど安全)')
ax.set_ylabel('地点数')
ax.set_title('Ji の分布 (balanced)')
fig.tight_layout()
fig.savefig(FIG / 'ji_histogram.png', dpi=150)
print('saved', FIG / 'ji_histogram.png')
```

## 図2: 寄与3項の平均構成

```python
dist = (n1 * df['d']).mean()
disc = (n2 * (100 - df['C'])).mean()
heat = (n3 * df['WBGT']).mean()
fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(['距離', '不快', '暑さ'], [dist, disc, heat],
       color=['#2a9d8f', '#f4813f', '#e63946'])
ax.set_ylabel('平均寄与')
ax.set_title('Ji への平均寄与 (balanced)')
fig.tight_layout()
fig.savefig(FIG / 'contribution_bar.png', dpi=150)
print('saved', FIG / 'contribution_bar.png')
```
